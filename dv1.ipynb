```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

sales = [20000, 25000, 30000, 65000]

plt.plot(months, sales)

plt.title("Monthly Sales")

plt.xlabel("Month"); plt.ylabel("Sales")

plt.show()
```
