```python
def F(x):
    return 40 + 5*x - 0.2*x**2

def dF(x):
    return 5 - 0.4*x

for x in range(1, 11):
    print(x, F(x), dF(x))

print("Critical point:", 12.5)
```

```python
def L(x):
    return 5*x - 0.2*x**2

x = np.linspace(0, 5, 1000)
y = L(x)

area = np.trapezoid(y, x)

print("Numerical integral:", area)
```

```python
X = df[
    ["Study_Hours",
     "Attendance",
     "Assignment_Score"]
].values

Y = df["Final_Score"].values

# Add a column of 1s for the intercept
X = np.column_stack((np.ones(len(X)), X))

# Calculate beta
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)
```
